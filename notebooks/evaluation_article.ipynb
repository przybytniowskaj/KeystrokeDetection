```python
import pandas as pd
data = pd.read_csv(f"../results/{model_name}.csv")
```

```python
stats_df = data.groupby(['Type', 'ID', 'Target Sentence'])[['Lev Dist', 'Gemini Lev Dist']].agg(['min', 'max', 'mean'])

stats_df.columns = ['_'.join(col).strip() for col in stats_df.columns.values]
stats_df = stats_df.reset_index()

stats_df.to_csv("../results/noisy_stats.csv", index=False)
```
